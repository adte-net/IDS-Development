```python
import os
import requests

# 1. Apply proxy variables explicitly
proxy_url = "http://172.16.1.11:3128"  # <--- Change to your Squid IP & Port
os.environ['http_proxy'] = proxy_url
os.environ['https_proxy'] = proxy_url
os.environ['no_proxy'] = "localhost,127.0.0.1,172.16.1.145"

print("Testing connection through proxy...")

try:
    # Test a simple outbound secure request
    response = requests.get("https://pypi.org", timeout=10)
    print(f"Success! Status Code: {response.status_code}")
except requests.exceptions.ProxyError as e:
    print(f"Proxy Connection Refused: {e}")
except requests.exceptions.Timeout:
    print("Connection Timed Out. Check if the proxy IP/Port is correct or if it blocks this domain.")
except Exception as e:
    print(f"Failed: {e}")
```
